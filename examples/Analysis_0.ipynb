```python
# Step 1: Install the kagglehub library (only need to run this once)
!pip install kagglehub

import kagglehub
import pandas as pd
import os

# Step 2: Download the dataset
print("Downloading dataset...")
path = kagglehub.dataset_download("mobeenfatimah/student-exam-performance-and-success-dataset")
print("Dataset downloaded to:", path)

# Step 3: Look inside the folder to find the actual CSV file name
files_in_folder = os.listdir(path)
print("Files found in the folder:", files_in_folder)

# Step 4: Automatically find the first .csv file and load it
csv_filename = [f for f in files_in_folder if f.endswith('.csv')][0]
full_csv_path = os.path.join(path, csv_filename)

print(f"Loading file: {csv_filename}")
df = pd.read_csv(full_csv_path)

# Step 5: Verify it worked
print("\nDataset loaded successfully!")
print(f"Shape of dataset: {df.shape}")
df.head()
```

```python
# 1. Install the library (only need to run this once per session)
!pip install ydata-profiling

# 2. Import the library
from ydata_profiling import ProfileReport

# 3. Generate the profile report
# We set explorative=True to get deeper insights into text and categorical data
print("Generating report... this may take a few seconds.")
profile = ProfileReport(df, title="Student Exam Performance & Success Profiling", explorative=True)

# 4. Display the interactive report directly inside your Colab notebook
profile.to_notebook_iframe()
```

### 💡 Key Business Insights & Recommendations

After analyzing the data quality, distributions, and correlations using `ydata-profiling`, here are three actionable insights for educational stakeholders:

#### 1. Study Habits Directly Impact Performance
The correlation analysis reveals that `study_hours_per_day` and `self_study_hours` are highly correlated, and both likely influence `exam_score`, `performance_grade`, and `pass_status`.

- **Insight:** Students who dedicate consistent time to studying (both structured and self-directed) achieve significantly better exam outcomes. The strong correlation between study time and performance metrics suggests that *time investment* is a critical success factor.
- **Action:** Educational institutions should implement **mandatory study hour programs** and provide structured self-study resources. Consider creating study schedules for at-risk students and tracking study time as an early-warning indicator.

#### 2. Critical Data Quality Issue: High Missing Values in Key Variables
The report reveals concerning data quality issues with **6-10% missing values** across critical variables:
- `attendance_percentage`: 9.9% missing
- `time_management_score`: 9.6% missing
- `notes_quality`: 8.4% missing
- `previous_gpa`: 7.8% missing

- **Insight:** Nearly 10% of students are not reporting attendance and time management data. This missing data is likely **non-random** - students who skip classes or struggle with time management may be avoiding these questions, creating a blind spot in identifying at-risk students.
- **Action:**
  1. **Improve survey design:** Make these fields mandatory or use automated tracking (e.g., LMS login data for attendance).
  2. **Use statistical imputation:** Fill missing values using median scores from similar student profiles.
  3. **Investigate the gap:** Conduct follow-up surveys with non-respondents to understand why they're not reporting.

#### 3. Mental Health Crisis: Stress and Anxiety are Interconnected
The report shows `stress_level` and `exam_anxiety_level` are **highly correlated** with each other.

- **Insight:** Student mental health is a significant factor in academic performance. High stress and exam anxiety likely create a negative feedback loop that impacts study effectiveness and exam outcomes.
- **Action:**
  - Implement **mental health support programs** including counseling services and stress management workshops.
  - Introduce **low-stakes assessments** throughout the semester to reduce exam anxiety.
  - Train faculty to recognize signs of stress and anxiety in students.

### 🛠️ Technical Notes on Data Quality
- **Missing Values:** Multiple columns have 6-10% missing data, requiring careful imputation strategies before modeling.
- **Imbalanced Variable:** `internet_access` shows 59.3% imbalance, which may introduce bias in analyses related to digital learning resources.

- **Previous Performance:** Strong correlation between `previous_gpa` and `previous_exam_score` suggests historical performance is a reliable predictor of current outcomes.
